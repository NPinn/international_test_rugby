```python
import sys
import os

import json
import pandas as pd
import numpy as np

_rel_path = os.path.realpath('')
base_path = os.path.abspath(_rel_path)

data_path = os.path.join(base_path, 'data')
raw_data_path = os.path.join(data_path, 'raw_data')
cleaned_data_path = os.path.join(data_path, 'cleaned_data')
```

```python
all_rugby_matches_dict = json.load(open(os.path.join(raw_data_path, 'all_rugby_matches.json')))
```

```python
international_teams_dict = json.load(open(os.path.join(raw_data_path, "international_rugby_team_ids.json")))
```

```python
all_rugby_matches_df = pd.DataFrame().from_dict(all_rugby_matches_dict).T.reset_index(names=['matchId'])

print(f'Number of Total Matches: {all_rugby_matches_df.shape[0]}\n')
```

```python
international_rugby_matches = all_rugby_matches_df[(all_rugby_matches_df['teamAid'].isin(international_teams_dict.keys()))&(all_rugby_matches_df['teamBid'].isin(international_teams_dict.keys()))]
international_rugby_matches = international_rugby_matches.reset_index(drop=True)

print(f'Number of International Matches: {international_rugby_matches.shape[0]}\n')

international_rugby_matches.head()
```

```python
def scoring_system(row):
    scoring_system = ''
    if row['matchDate'] <= '1875-12-31':
        scoring_system = '1'
    elif row['matchDate'] >= '1876-01-01' and row['matchDate'] <= '1889-12-31':
        scoring_system = '2'
    elif row['matchDate'] >= '1890-01-01' and row['matchDate'] <= '1891-12-31':
        scoring_system = '3'
    elif row['matchDate'] >= '1892-01-01' and row['matchDate'] <= '1893-12-31':
        scoring_system = '4'
    elif row['matchDate'] >= '1894-01-01' and row['matchDate'] <= '1904-12-31':
        scoring_system = '5'
    elif row['matchDate'] >= '1905-01-01' and row['matchDate'] <= '1947-12-31':
        scoring_system = '6'
    elif row['matchDate'] >= '1948-01-01' and row['matchDate'] <= '1970-12-31':
        scoring_system = '7'
    elif row['matchDate'] >= '1971-01-01' and row['matchDate'] <= '1971-12-31':
        scoring_system = '8'
    elif row['matchDate'] >= '1972-01-01' and row['matchDate'] <= '1991-12-31':
        scoring_system = '9'
    elif row['matchDate'] >= '1992-01-01':
        scoring_system = '10'

    return scoring_system
```

```python
international_rugby_matches['scoring_system'] = international_rugby_matches.apply(lambda x: scoring_system(x), axis=1)
```

```python
international_rugby_matches['matchDate'] = pd.to_datetime(international_rugby_matches['matchDate'])
international_rugby_matches = international_rugby_matches.sort_values(by='matchDate').reset_index(drop=True)

international_rugby_matches.head()
```
