```python
import requests
import json
import os
import glob
from credentials import *

_rel_path = os.path.realpath('')
base_path = os.path.abspath(_rel_path)

data_path = os.path.join(base_path, 'data')
raw_data_path = os.path.join(data_path, 'raw_data')
cleaned_data_path = os.path.join(data_path, 'cleaned_data')
```

```python
all_rugby_matches_dict = json.load(open(os.path.join(raw_data_path, 'all_rugby_matches.json')))
```

```python
international_teams_dict = json.load(open(os.path.join(raw_data_path, "international_rugby_team_ids.json")))
```

```python
print('Extracting Details')
page = 0
all_matches = []

while True:
    ret = requests.get(f"{WR_API}/match?pageSize=100&sort=asc&page={page}").json()

    matches = ret['content']

    if not matches:
        break

    print(f"Results as of Page {page}: Retrieved {len(matches)} items (total so far: {len(all_matches)})")

    for i in range(len(matches)):
        if matches[i]['matchId'] not in all_rugby_matches_dict.keys() and\
           matches[i]['teams'][0]['id'] in international_teams_dict.keys() and\
           matches[i]['teams'][1]['id'] in international_teams_dict.keys():
            all_matches.append(matches[i])
        else:
            pass

    page += 1
print(f"Total matches retrieved: {len(all_matches)}")
```
